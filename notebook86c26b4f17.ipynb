```python
import ast
```

```python
def convert(obj): #string obj obtained
    L = []
    for i in ast.literal_eval(obj): #loop runs on list
        L.append(i['name']) 
    return L 
```

```python
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
        counter+=1
    return L 
```

#### NLTK

There may occur different forms of teh same root word in our vector list which we don't word since the root word is sufficient so we use the nltk lib to do stemming which trims the words of different forms to their root form....  
That way we can also add more words in our bag or vector list  
Ex: love,loving,loved indicates the same root word love

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer() #Porter stemming algorithm is a process for removing the commoner morphological and inflexional endings from words in English. Its main use is as part of a term normalisation process that is usually done when setting up Information Retrieval systems.
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
ps.stem('running')
```

```python
new_df['tags']=new_df['tags'].apply(stem)
```

### Text Vectorization

Bag of words approach...<br>
-> A bag of w words are chosen/taken<br>
-> Each movie is analysed to see how many times it has each of those w words occuring  
-> This forms a vector of each of the movies  
-> We then take the k nearest neighbors of the movie we searched for to get the recommendations similar to that movie in that vector space  



```python
# we won't be using stop words as they dont contribute to the semantic meaning of the sentence...they are just for support in sentence formation
from sklearn.feature_extraction.text import CountVectorizer #count vectorizer is a class to convert a collection of text documents to a matrix of token counts. It tokenizes the text, builds a vocabulary of known words, and encodes new documents using that vocabulary.
cv = CountVectorizer(max_features=5000,stop_words='english') #hyperparameter max_features=5000 means that we will only consider the top 5000 most frequent words in the dataset. This is done to reduce the dimensionality of the feature space and to focus on the most important words that are likely to contribute to the meaning of the text. The stop_words='english' parameter means that we will ignore common English stop words (such as "the", "and", "is", etc.) when creating the feature matrix, as these words do not carry much meaning and can add noise to the data.
    
```

```python
vector = cv.fit_transform(new_df['tags']).toarray() #bydef cv returns a scipy sparse matrix but we need a dense matrix so we convert it to a numpy array using toarray() method. The fit_transform() method of the CountVectorizer class is used to learn the vocabulary of the input data and transform it into a document-term matrix, where each row represents a document and each column represents a unique word in the vocabulary. The resulting matrix contains the frequency counts of each word in each document.
```

Sparse matrix → mostly zero values. Dense matrix → mostly non-zero values.

Sparse representation → stores non-zero values and information about their positions.  
Dense representation → stores every value, including zeros; positions are implied by the array structure.

here cv returns a sparse matrix and is fine to work with cosine similarity  
converting to dense matrix(np arr)is optional

--We need to calculate the distance between each vectors(movies) which would be the cosine(angular) distance and not the euclidean distance.  
--If angle=0 then same movie, if angle very less then similar movies, if angle = 90deg then huge dissimilarity and if angle=180deg then totally dissimilar or opposite movies.  
--Note: Euc dist is not a good measure in higher dimensions due to curse of dimensionality hence for higher dim. calc we prefer cosine distance.  
--similarity=1/distance (0 to 1)  
0:less sim more dist and 1:less dist more sim

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity.shape
```

```python
similarity #array of arrays
```

```python
similarity[0] #every movie is similar to itself and hence the first element of the first array is 1. The other elements of the first array represent the similarity of the first movie with all other movies in the dataset, with values ranging from 0 to 1, where 0 indicates no similarity and 1 indicates perfect similarity. hence the diagonal of the matrix would be 1 and the rest of the elements would be between 0 and 1. The similarity values are calculated based on the cosine similarity between the feature vectors of the movies, which are derived from their tags.
```

Flow of recommendation:  
input : movie name -> get that movie's idx -> hit the similarity(idx) fn for that movie idx -> sort the list/array -> fetch the top 5 most similar movies

```python
new_df[new_df['title'] == 'The Lego Movie'].index[0]
# new_df[new_df['title'] == 'The Lego Movie']
```

```python
def recommend(movie):
    index = new_df[new_df['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new_df.iloc[i[0]].title)
        
    
```

```python
recommend('Batman Begins')
```
